```python
# Singly Linked List
class LlNode:
  def __init__(self, input_data=None):
    self.data = input_data
    self.next = None

class Ll:
  def __init__(self):
    self.head = None

  def insert(self, new_node):
    """insert new_node at the head
    O(1) """
    new_node.next = self.head
    self.head = new_node

  def delete(self, del_data):
    """ Remove node with del_data from linked list, leaving linked list intact!
    Returns True if successful,
    returns False if del_data not in any nodes."""
    found = self.find(del_data)
    #if not found:
    #if found == None:
    if found == None:
      return False
    else:


      return True

  def find(self, find_data):
    """ cycles through nodes looking for one with data
    equal to find data;
    returns the first node with find_data, otherwise
    returns None
    O(n) """
    cur_node = self.head
    while cur_node:
      if cur_node.data == find_data:
        return cur_node
      cur_node = cur_node.next
    # we never found the find_data
    return None

  def print_all(self):
    """ O(n) """
    cur_node = self.head
    while cur_node:
      print(cur_node.data)
      cur_node = cur_node.next

ll = Ll() # create a linked list

ln1 = LlNode('A')
ll.insert(ln1)
ln2 = LlNode('B')
ll.insert(ln2)
ll.print_all()

found_node = ll.find('B')
print(found_node)
if found_node != None:
  print(found_node.data)
```
