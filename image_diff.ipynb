### Image manipulation
In the section below, we overlay one image onto another. This is one of the most common ways to merge and manipulate images.

```python
!wget "https://raw.githubusercontent.com/taylan-sen/NUSTEM_SUMMER26/refs/heads/main/mystery.png"
```

```python
mystery = cv2.resize(cv2.imread("mystery.png",0),(130,150))
print(mystery.shape)
```

```python
# create a new image trojanhorse2, which is the combination of
# trojanhorse1 and the mystery image
trojanhorse2 = img_gray0.copy()
trojanhorse2[0:150,0:130] = np.minimum(mystery,trojanhorse2[0:150,0:130])
plt.imshow(trojanhorse2,cmap="gray")
```

### COLOR IMAGES
The section below shows how color images are stored using a separate matrix for red green and blue.

![](https://images.timesnownews.com/thumb/msid-119004036,thumbsize-102632,width-1280,height-720,resizemode-75/119004036.jpg)

```python
!wget "https://images.timesnownews.com/thumb/msid-119004036,thumbsize-102632,width-1280,height-720,resizemode-75/119004036.jpg" -O image1.jpg
```

```python
# LOAD A COLOR IMAGE
img_bgr = cv2.imread("image1.jpg")
print(img_bgr.shape)
```

```python
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
plt.imshow(img_bgr)
plt.show()

plt.imshow(img_rgb)
plt.show()
```

Notice above how the first image's colors are distorted. This highlights how some computer systems use different color formats (i.e. BGR vs. RGB).

```python
# The following code shows the color decomposition of an image
img_blue = img_rgb.copy()
img_blue[:,:,0:2] = 0
plt.imshow(img_blue)
plt.show()

img_red = img_rgb.copy()
img_red[:,:,1:3] = 0
plt.imshow(img_red)
plt.show()

img_green = img_rgb.copy()
img_green[:,:,0] = 0
img_green[:,:,2] = 0
plt.imshow(img_green)
plt.show()
```
